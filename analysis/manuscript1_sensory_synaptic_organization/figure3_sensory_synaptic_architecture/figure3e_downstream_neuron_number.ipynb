# Figure 3e — Number of downstream neurons per sensory neuron

Number of distinct, intact spinal cord neurons postsynaptic to each reconstructed sensory
neuron.

**Method.** Start from the sensory neuron -> postsynaptic partner table
(`data/sensory_postsynaptic_partners.csv`: one row per pre/post pair, with the
postsynaptic neuron's supervoxel count and the synapse count between them). Keep only
partners that are reasonably completely reconstructed — `num_supervoxels >= 40000` — then
count distinct postsynaptic partners per sensory neuron. Sensory neurons with no intact
partner contribute a zero.

**Statistics.** Kruskal-Wallis followed by Dunn post hoc (Holm correction).

This panel is the one place in Figure 3 that reads a shipped table rather than querying
CAVE directly: the partner table carries a per-neuron `num_supervoxels` value that is not
available from the synapse table alone.

```python
# Internal subtype key -> published label
SUBTYPE_LABELS = {
    "c-ltmr":  "C-LTMR",
    "mrgd":    "C-HTMR/Heat (MrgprA3,MrgprD,MrgprB4)",
    "ad-ltmr": "Aδ-LTMR",
    "ab-ltmr": "Aβ-LTMR (single collateral)",
    "sst":     "C-HTMR/Heat (SST)",
    "ad-htmr": "Aδ-HTMR (single collateral)",
    "chtmr-p": "Peptidergic C-Nociceptors",
    "c-cold":  "C-Cold",
}

PALETTE = {
    "C-LTMR":                               "#BCEBCB",
    "C-HTMR/Heat (MrgprA3,MrgprD,MrgprB4)": "#b23aee",
    "Aδ-LTMR":                              "#00A86B",
    "Aβ-LTMR (single collateral)":          "#556B2F",
    "C-HTMR/Heat (SST)":                    "#ffbbff",
    "Aδ-HTMR (single collateral)":          "#ff4040",
    "Peptidergic C-Nociceptors":            "#ffee00",
    "C-Cold":                               "#00bfff",
}
```

```python
from pathlib import Path

PARTNERS_CSV = Path("../data/sensory_postsynaptic_partners.csv")
MIN_SUPERVOXELS = 40_000   # "intact" postsynaptic neuron threshold
```

```python
import numpy as np
import pandas as pd

id_to_subtype = {rid: st for st, ids in SENSORY_NEURONS.items() for rid in ids}

df = pd.read_csv(PARTNERS_CSV)
df = df.dropna(subset=["pre_pt_root_id", "post_pt_root_id", "num_supervoxels"])
df["pre_pt_root_id"] = df["pre_pt_root_id"].astype(np.int64)
df["post_pt_root_id"] = df["post_pt_root_id"].astype(np.int64)

# sensory presynaptic neurons only, intact postsynaptic partners only
df = df[df["pre_pt_root_id"].isin(id_to_subtype)]
df = df[df["num_supervoxels"] >= MIN_SUPERVOXELS]

counts = (df.groupby("pre_pt_root_id")["post_pt_root_id"]
            .nunique()
            .reindex(list(id_to_subtype))
            .fillna(0)
            .astype(int)
            .rename("Downstream Neurons")
            .reset_index())

counts["Neuron Type"] = counts["pre_pt_root_id"].map(id_to_subtype)
counts["Neuron Label"] = counts["Neuron Type"].map(SUBTYPE_LABELS)

print(counts.groupby("Neuron Label")["Downstream Neurons"]
            .agg(n="size", mean="mean", sd="std", median="median"))
```

```python
import numpy as np
import pandas as pd
import scikit_posthocs as sp
from scipy.stats import kruskal


def kruskal_dunn(df, value_col, group_col="Neuron Label"):
    """Kruskal-Wallis across subtypes, then Dunn post hoc with Holm correction."""
    groups = [g[value_col].to_numpy(float) for _, g in df.groupby(group_col)]
    H, p = kruskal(*groups)
    print(f"Kruskal-Wallis: H = {H:.4g}, p = {p:.4g}")

    dunn = sp.posthoc_dunn(df, val_col=value_col, group_col=group_col,
                           p_adjust="holm")
    return dunn


def bar_with_points(df, value_col, ylabel, order, figsize, yticks=None,
                    out_svg=None, seed=0):
    """Bar of subtype means with individual neurons overlaid as jittered points.

    Deliberately plain matplotlib: seaborn's barplot error-bar argument changed
    name across 0.11 / 0.12 (`ci=` -> `errorbar=`), so a seaborn version would
    break depending on which environment the notebook is run in.
    """
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=figsize)
    x = np.arange(len(order))

    means = [df.loc[df["Neuron Label"] == g, value_col].mean() for g in order]
    ax.bar(x, means, width=0.8, color=[PALETTE[g] for g in order],
           edgecolor="none", zorder=1)

    rng = np.random.default_rng(seed)
    for i, g in enumerate(order):
        vals = df.loc[df["Neuron Label"] == g, value_col].to_numpy(float)
        ax.scatter(i + rng.uniform(-0.2, 0.2, len(vals)), vals,
                   s=4, color="black", alpha=0.4, linewidths=0, zorder=3)

    ax.set_xticks(x)
    ax.set_xticklabels(order, rotation=-50, ha="left", fontsize=4)
    ax.set_xlim(-0.6, len(order) - 0.4)
    if yticks is not None:
        ax.set_yticks(yticks)
    ax.set_ylabel(ylabel, fontsize=5)
    ax.tick_params(axis="y", labelsize=5)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.grid(False)
    fig.tight_layout()

    if out_svg:
        fig.savefig(out_svg, dpi=600, bbox_inches="tight")
    plt.show()
```
